```python
import os
```

```python
os.getcwd()
```

```python
output_folder = "cleaned_data"

os.makedirs(output_folder, exist_ok=True)
```

```python
os.listdir()
```

```python
datasets = {
    "dim_date_cleaned.csv": date,
    "dim_product_cleaned.csv": product,
    "dim_store_reviewed.csv": store,
    "fact_daily_sales_reviewed.csv": sales,
    "fact_store_monthly_profitability_reviewed.csv": fact
}
```

```python
for filename, dataframe in datasets.items():

    file_path = os.path.join(output_folder, filename)

    dataframe.to_csv(file_path, index=False)

    print(filename, "saved successfully")
```

```python
os.listdir(output_folder)
```
